```python
from bs4 import BeautifulSoup
```

```python
import pandas as pd
import re
```

```python
with open("movies_html.html", "r", encoding="utf-8") as f:
    html = f.read()

soup = BeautifulSoup(html, "lxml")
```

```python
title = soup.find("h1").text
print(title)
```

```python
try:
        genre_label = soup.find(string="Genres")

        if genre_label:
            genre_container = genre_label.find_next("div", class_="poster-detail-infos__value")
            genre = genre_container.find("span").text.strip()
        else:
            genre = None

    except:
        genre = None 
    genre = genre.split(",")[0] if genre else None #keep first genre only
```

```python
movies = html.split("<!-- MOVIE")

urls=[]
titles = []
years = []
imdb_ratings = []
imdb_reviews_list = []
runtime_list = []
directors_list = []
country_list = []
genres_list = []
main_actors = []



for movie_html in movies[1:]:

    soup = BeautifulSoup(movie_html, "html.parser")
    
    try:
        title = soup.find("h1").text.strip().split("(")[0].strip()
    except:
        title = None
    
    try: 
        year = soup.find('span',class_='release-year').text.strip().replace("(", "").replace(")", "")
    except:
        year = None

    try:
        imdb_tag = soup.find('span', class_="imdb-score")
        imdb_text = imdb_tag.text.strip()

        imdb_rating = imdb_text.split()[0]
        imdb_review_count = imdb_text.split()[1].replace("(", "").replace(")", "")

    except:
        imdb_rating = None
        imdb_review_count = None

    try:
        items = soup.find_all('div', class_='title-detail-hero-details__item')
    
        runtime = None

        for item in items:
            text = item.text.strip()
        
            if 'min' in text:   # runtime always contains 'min'
               runtime = text
               break

    except:
         runtime = None

    # extract URL safely
    try:
        url_line = movie_html.split("<!-- URL: ")[1].split("-->")[0].strip()
    except:
        url_line = None 
    
    try:
        director = soup.find('span', class_='title-credit-name').text.strip()
    
    except:
        director = None

    try:
        country_label = soup.find(string="Production country")

        if country_label:
            country_text = country_label.find_next("div", class_="poster-detail-infos__value").text.strip()

            primary_country = country_text.split(",")[0].strip()
        else:
            primary_country = None

    except:
        primary_country = None

    try:
        genre_label = soup.find(string="Genres")

        if genre_label:
            genre_container = genre_label.find_next("div", class_="poster-detail-infos__value")
            genre = genre_container.find("span").text.strip()
        else:
            genre = None

    except:
        genre = None 
    genre = genre.split(",")[0] if genre else None #keep first genre only  

    try:
        actor_tags = soup.find_all("span", class_="title-credit-name")

        if actor_tags and len(actor_tags) > 1:
            main_actor = actor_tags[1].text.strip()   # first actor
        else:
            main_actor = None

    except:
        main_actor = None

    runtime_list.append(runtime)
    imdb_ratings.append(imdb_rating)
    imdb_reviews_list.append(imdb_review_count)
    titles.append(title)
    years.append(year)
    urls.append(url_line)
    directors_list.append(director)
    country_list.append(primary_country)
    genres_list.append(genre)
    main_actors.append(main_actor)
   
    
    
print(titles, years, imdb_ratings, imdb_reviews_list,runtime_list,urls,directors_list,country_list,genres_list,main_actors)
```

```python
df = pd.DataFrame({
    "title": titles,
    "year": years,
    'genre':genres_list,
    "runtime":runtime_list,
    "imdb_rating": imdb_ratings,
    "country":country_list,
    "director":directors_list,
    "no._of_reviews": imdb_reviews_list,
    "url": urls,
    'actor':main_actors
})

df
```

```python
#change runtime hour and minute into only minutes 
def runtime_to_minutes(text):
    if text is None:
        return None
    
    hours = re.search(r'(\d+)h', text)
    mins = re.search(r'(\d+)min', text)
    
    h = int(hours.group(1)) if hours else 0
    m = int(mins.group(1)) if mins else 0
    
    return h * 60 + m
```

```python
df['runtime'] = df['runtime'].apply(runtime_to_minutes)
```

```python
df.rename(columns={'runtime':'runtime_in_min'},inplace=True)
```

```python
#convert reviews no. from k or m to number
def convert_reviews(x):
    if isinstance(x, str):
        x = x.lower()
        if "k" in x:
            return float(x.replace("k","")) * 1000
        elif "m" in x:
            return float(x.replace("m","")) * 1000000
        else:
            return float(x)
    return x

df["no._of_reviews"] = df["no._of_reviews"].apply(convert_reviews)
```
